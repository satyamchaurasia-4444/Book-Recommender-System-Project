```python
popular_df=avg_rating.merge(num_rating,on='Book-Title')
```

Collabrative based recommendation system

```python
x=rating_with_name.groupby('User-ID').count()['Book-Rating']>200
```

```python
padhe_likhe_users=x[x].index
```

```python
filtered_rating=rating_with_name[rating_with_name['User-ID'].isin(padhe_likhe_users)]
```

```python
y=filtered_rating.groupby('Book-Title').count()['Book-Rating']>50
```

```python
famous_books=y[y].index
```

```python
famous_books
```

```python
filtered_rating=filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
```

```python
filtered_rating.head()
```

```python
pt=filtered_rating.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
```

```python
pt=pt.fillna('0')
pt
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(pt)
```

```python
similarity
```

```python
np.where(pt.index == '1984')[0][0]
```

```python
def book_recommender(book):
    index = np.where(pt.index == book)[0][0]
    similarity[index]
    df=pd.DataFrame({
    'book':pt.index,
    'score':similarity[index]
    })
    r_books=df.sort_values('score',ascending=False)[1:6]
    r_books=list(r_books['book'])
    return r_books
```

```python
book_recommender('1984')
```
